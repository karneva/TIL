# BFS (Breadth First Search)

## **1. 개념**

- **너비 우선 탐색(BFS)**은 **시작 노드**로부터 가까운 노드를 먼저 방문하고, 다음으로 그보다 한 단계 더 멀리 있는 노드를 차례대로 방문하는 탐색 방식
- 자료구조 중 큐(Queue)의 **선입선출(FIFO)** 특성과 잘 맞는다
- 노드를 방문하면, **인접한 모든 노드**를 큐에 넣어(삽입) **가장 먼저 들어온 노드부터** 차례대로 탐색을 진행

## **2. BFS 구현 방식**

BFS는 보통 **인접 행렬(Adjacency Matrix)** 혹은 인접 리스트(Adjacency List)를 사용해 **큐**를 통해 구현.

아래는 **인접 행렬 + 큐**와 **인접 리스트 + 큐**를 예시로 소개.

- 아래 2가지 예시 코드는 다음 그래프 정보를 받습니다.
    - input & output
        
        ```python
        7 8
        1 2 1 3 2 4 2 5 4 6 5 6 6 7 3 7
        ```
        
        - 총 8개의 간선: (1,2), (1,3), (2,4), (2,5), (4,6), (5,6), (6,7), (3,7)
        
        ```python
        1234576
        ```
        
    - BFS 시작 노드를 1이라 할 때, 탐색 결과를 순서대로 출력하면:**`1234576`**
    - 1에 인접한 (2,3)를 먼저 방문 → 2를 방문하면서 (4,5) → 3을 방문하면서 (7) … 차례대로 큐에서 처리

### 2.1 인접 행렬 + 큐

````python
V, E = map(int, input().split())
data = list(map(int, input().split()))

# 인접 행렬 생성
adj_matrix = [[0] * (V + 1) for _ in range(V + 1)]

# 간선 정보를 인접 행렬에 기록 (무방향 그래프)
for i in range(E):
    n1, n2 = data[2 * i], data[2 * i + 1]
    adj_matrix[n1][n2] = 1
    adj_matrix[n2][n1] = 1

def BFS_matrix(start):
    visited = [0] * (V + 1)  # 방문 여부 리스트
    q = []  # 큐 초기화

    # 시작 노드 방문 처리 후 큐에 삽입 (enqueue)
    # 작성 순서는 상관 없지만 ‘방문 처리 → 큐에 삽입’ 순서가 조금 더 자연스러움
    # “큐에 넣었다”는 것은 “이 노드를 이미 발견(visited)해서 다음에 처리할 예정”이라는 의미
    visited[start] = 1
    q.append(start)

    while q:
        current_node = q.pop(0)  # 리스트 맨 앞에서 원소 제거 (dequeue)
        print(current_node, end=' ')

        # current_node와 인접한 노드 탐색
        for next_node in range(1, V + 1):
            # 연결되어 있고, 아직 방문 안 했다면 => 방문 처리 후 큐에 삽입 (enqueue)
            if (
                adj_matrix[current_node][next_node] == 1
                and visited[next_node] == 0
            ):
                visited[next_node] = 1
                q.append(next_node)

# BFS 실행
BFS_matrix(1)
````

### 2.2 인접 리스트 + 큐

````python
V, E = map(int, input().split())
data = list(map(int, input().split()))

# 인접 리스트 생성
adj_list = [[] for _ in range(V + 1)]

# 간선 정보를 인접 리스트에 기록 (무방향 그래프)
for i in range(E):
    n1, n2 = data[2 * i], data[2 * i + 1]
    adj_list[n1].append(n2)
    adj_list[n2].append(n1)

# 방문 시, 작은 번호부터 접근하기 위해 각 간선정보 리스트를 오름차순 정렬
for i in range(1, V + 1):
    adj_list[i].sort()

def BFS_list(start):
    visited = [0] * (V + 1)  # 방문 여부 리스트
    q = []  # 큐 초기화

    # 초기 설정: 시작 노드 방문 후 큐에 삽입
    visited[start] = 1
    q.append(start)

    while q:
        current_node = q.pop(0)  # 리스트 맨 앞에서 원소 제거
        print(current_node, end=' ')

        # current_node와 인접한 노드 탐색
        for next_node in adj_list[current_node]:
            # 방문 안 했다면 => 방문 후 큐에 삽입
            if visited[next_node] == 0:
                visited[next_node] = 1
                q.append(next_node)

# BFS 실행
BFS_list(1)
````

### 2.3 Deque를 활용한 예시

````python
# 인접 행렬 + Deque
from collections import deque

V, E = map(int, input().split())
data = list(map(int, input().split()))

adj_matrix = [[0] * (V + 1) for _ in range(V + 1)]

for i in range(E):
    n1, n2 = data[2 * i], data[2 * i + 1]
    adj_matrix[n1][n2] = 1
    adj_matrix[n2][n1] = 1

def BFS_matrix(start):
    visited = [0] * (V + 1)
    q = deque()

    visited[start] = 1
    q.append(start)

    while q:
        current_node = q.popleft()
        print(current_node, end='')

        for next_node in range(1, V + 1):

            if (
                adj_matrix[current_node][next_node] == 1
                and visited[next_node] == 0
            ):
                visited[next_node] = 1
                q.append(next_node)

BFS_matrix(1)
````

````python
# 인접 리스트 + Deque
from collections import deque

V, E = map(int, input().split())
data = list(map(int, input().split()))

adj_list = [[] for _ in range(V + 1)]

for i in range(E):
    n1, n2 = data[2 * i], data[2 * i + 1]
    adj_list[n1].append(n2)
    adj_list[n2].append(n1)

for i in range(1, V + 1):
    adj_list[i].sort()

def BFS_list(start):
    visited = [0] * (V + 1)
    q = deque()

    visited[start] = 1
    q.append(start)

    while q:
        current = q.popleft()
        print(current, end='')
        for next_node in adj_list[current]:
            if visited[next_node] == 0:
                visited[next_node] = 1
                q.append(next_node)

BFS_list(1)
````

## **3. BFS 알고리즘 핵심 순서**

1. **시작 노드**를 방문 표시 후 **큐**에 삽입
2. **큐가 빌 때까지** 반복:
    - 큐에서 노드를 꺼냄(`pop(0)`)
    - 해당 노드의 **인접 노드들** 중 아직 방문하지 않은 노드를 **전부 방문 표시**하고 **큐에 삽입**
3. 큐가 공백이 되면(더 이상 방문할 노드가 없으면) BFS 종료

> FIFO 자료구조인 큐를 사용함으로써, 가까운 노드부터 차례대로 처리하게 됨

## **4. DFS와의 비교**

- **DFS (깊이 우선 탐색)**
    - 한 경로로 **깊이** 들어갔다가 더 이상 갈 곳이 없으면 뒤로 돌아옴(백트래킹)
    - **스택(Stack)** 또는 **재귀**(호출 스택) 사용
    - 순차적으로 **깊은 경로**를 먼저 탐색
- **BFS (너비 우선 탐색)**
    - **시작 노드에서 가까운 노드**부터 차례로 방문, 다음으로 한 단계 더 멀리 있는 노드를 방문
    - **큐(Queue)** 사용
    - **넓은 범위**를 레벨 순으로 탐색

> “BFS는 스택으로 구현할 수 없다”와 같은 말은 엄밀히 틀림.
> 
> 
> 추가 연산을 통해 스택으로 구현해도 되지만, 
> **가장 자연스럽고 효율적인 방법**은 `BFS ↔ 큐`, `DFS ↔ 스택(재귀)`이다.

## **5. 마무리**

- **BFS(너비 우선 탐색)**:
    1. **인접 행렬 + 큐(리스트 구현)**
    2. **인접 리스트 + 큐(리스트 구현)**
    3. (다른 그래프 표현도 가능)
- **핵심**: 큐에 삽입된 노드를 순차적으로 꺼내면서, **아직 방문하지 않은 인접 노드**를 **새롭게 방문**하고 큐에 삽입
- **DFS와의 차이**:
    - DFS는 **한 방향으로 깊이**, BFS는 **가까운 노드부터 레벨 순**
    - DFS ↔ 스택(재귀), BFS ↔ 큐가 가장 자연스럽고 효율적인 자료구조 매칭
